```python
import matplotlib.pyplot as plt

category_sales = df.groupby('Category')['Total_Sales'].sum().sort_values(ascending=False)

category_sales.plot(kind='bar', figsize=(8, 5))

plt.title('Total Sales by Category')
plt.xlabel('Category')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
region_sales = df.groupby('Region')['Total_Sales'].sum().sort_values(ascending=False)

region_sales.plot(kind='bar', figsize=(8, 5))

plt.title('Total Sales by Region')
plt.xlabel('Region')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
top_products = df.groupby('Product')['Total_Sales'].sum().sort_values(ascending=False).head(10)

top_products.plot(kind='bar', figsize=(10, 5))

plt.title('Top 10 Products by Sales')
plt.xlabel('Product')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df['Date'] = pd.to_datetime(df['Date'])
```

```python
df.dtypes
```

```python
monthly_sales = df.groupby(df['Date'].dt.to_period('M'))['Total_Sales'].sum()

monthly_sales
```

```python
monthly_sales.plot(kind='line', figsize=(10, 5))

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
best_region = df.groupby('Region')['Total_Sales'].sum().idxmax()
best_region_sales = df.groupby('Region')['Total_Sales'].sum().max()

print("Best Region:", best_region)
print("Sales:", best_region_sales)
```

```python
best_product = df.groupby('Product')['Total_Sales'].sum().idxmax()
best_product_sales = df.groupby('Product')['Total_Sales'].sum().max()

print("Best Product:", best_product)
print("Sales:", best_product_sales)
```

```python
# Final Business Insights

category_sales = df.groupby('Category')['Total_Sales'].sum().sort_values(ascending=False)
region_sales = df.groupby('Region')['Total_Sales'].sum().sort_values(ascending=False)
product_sales = df.groupby('Product')['Total_Sales'].sum().sort_values(ascending=False)

print("========== BUSINESS INSIGHTS ==========")

print("\n1. Best Performing Category:")
print(category_sales.index[0], "-", category_sales.iloc[0])

print("\n2. Best Performing Region:")
print(region_sales.index[0], "-", region_sales.iloc[0])

print("\n3. Best Selling Product by Revenue:")
print(product_sales.index[0], "-", product_sales.iloc[0])

print("\n4. Highest Sales Month:")
print(monthly_sales.idxmax(), "-", monthly_sales.max())

print("\n5. Total Sales:")
print(df['Total_Sales'].sum())

print("\n6. Average Sales per Transaction:")
print(df['Total_Sales'].mean())

print("\n7. Total Quantity Sold:")
print(df['Quantity'].sum())
```

## Conclusion

From this analysis, I was able to get a better understanding of the retail sales data and identify some useful sales patterns. I looked at the data based on categories, regions, products, and monthly sales to understand where the business is performing well.

The analysis helped me identify the categories and regions that contribute the most to sales. I also found the products that generate higher revenue and looked at how sales change from month to month. The correlation analysis gave me an idea of how variables such as quantity, unit price, and total sales are related.

Overall, the analysis shows that sales performance is not the same across all products and regions. Some areas perform better than others, so focusing on these differences can help the business make better decisions.

## Key Insights

* Some product categories contribute more to total sales than others.
* Certain regions have better sales performance compared to the rest.
* A few products contribute a significant amount of revenue.
* Monthly sales show changes in customer demand over time.
* Quantity, unit price, and total sales have relationships that can be understood through correlation analysis.

## Business Recommendations

1. The business should focus more on the categories and products that generate higher sales.

2. Regions with strong sales can be given more attention through targeted promotions and better stock availability.

3. Products with lower sales should be analyzed to understand whether pricing, demand, or marketing is affecting their performance.

4. The monthly sales pattern can be used to plan inventory and promotional campaigns during periods when customer demand is higher.

5. The business should regularly analyze sales data so that changes in customer preferences can be identified early.

6. Overall, using sales data for decision-making can help the business improve its marketing, inventory planning, and revenue performance.
