## Missing Data Handling Decisions

Different strategies are used depending on the meaning of each column.

### Age
Age is a numerical variable. The median is used for missing values because
median imputation is less affected by extreme values than mean imputation.

### Quantity
Quantity is a numerical sales variable. The median is used because it
represents a typical transaction quantity and avoids excessive influence
from unusually large values.

### Total_Sales
Total_Sales can be calculated from Quantity × Unit_Price. Therefore,
instead of using statistical imputation, missing Total_Sales values are
recalculated from the available sales information.

### Email
Email is not required for numerical analysis and cannot be reliably inferred.
Missing email values are therefore retained as missing rather than inventing
customer information.

### Gender
There are no missing gender values after standardisation, so no imputation
is required.

### Join_Date
There are no invalid dates after conversion. Invalid dates, if present,
would be converted to NaT and investigated rather than silently replaced.

```python
# Age: median imputation
df["Age"] = df["Age"].fillna(df["Age"].median())

# Quantity: median imputation
df["Quantity"] = df["Quantity"].fillna(df["Quantity"].median())

# Total_Sales: calculate from Quantity * Unit_Price
df["Total_Sales"] = df["Total_Sales"].fillna(
    df["Quantity"] * df["Unit_Price"]
)

print("Missing values after imputation:")
print(df.isnull().sum())
```

```python
duplicates_removed = df.duplicated().sum()

df = df.drop_duplicates().reset_index(drop=True)

print("Duplicate rows removed:", duplicates_removed)
print("Rows after duplicate removal:", len(df))
```

```python
def detect_iqr_outliers(dataframe, column):
    Q1 = dataframe[column].quantile(0.25)
    Q3 = dataframe[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = dataframe[
        (dataframe[column] < lower_bound) |
        (dataframe[column] > upper_bound)
    ]

    return outliers, lower_bound, upper_bound
```

```python
numeric_columns = ["Age", "Quantity", "Unit_Price", "Total_Sales"]

for column in numeric_columns:
    outliers, lower, upper = detect_iqr_outliers(df, column)

    print(f"\nColumn: {column}")
    print(f"Lower bound: {lower:.2f}")
    print(f"Upper bound: {upper:.2f}")
    print(f"Number of outliers: {len(outliers)}")
```

```python
# Replace impossible ages with NaN
df.loc[(df["Age"] < 18) | (df["Age"] > 100), "Age"] = np.nan

# Impute invalid ages with median
df["Age"] = df["Age"].fillna(df["Age"].median())

print("Age range after cleaning:")
print(df["Age"].min(), "to", df["Age"].max())
```

```python
df["Calculated_Sales"] = df["Quantity"] * df["Unit_Price"]

df["Sales_Match"] = np.isclose(
    df["Total_Sales"],
    df["Calculated_Sales"]
)

print("Incorrect sales records:",
      (~df["Sales_Match"]).sum())
```
